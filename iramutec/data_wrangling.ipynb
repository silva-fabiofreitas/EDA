```python
import pandas as pd

def format_for_iramuteq(df, text_column, metadata_columns=None, output_file='corpus_iramuteq.txt'):
    """
    Converte um DataFrame em formato compatível com IRaMuTeQ.

    Parâmetros:
    - df: DataFrame com os dados.
    - text_column: nome da coluna com os textos (respostas abertas, entrevistas...).
    - metadata_columns: lista de colunas com variáveis categóricas (opcional).
    - output_file: nome do arquivo de saída (.txt).
    """
    with open(output_file, 'w', encoding='utf-8') as f:
        for _, row in df.iterrows():
            # Início de uma nova unidade de contexto (UC)
            f.write("**** ")
            # Escreve as variáveis categóricas
            if metadata_columns:
                for col in metadata_columns:
                    value = str(row[col]).replace(" ", "_")  # IRaMuTeQ não aceita espaços em valores
                    f.write(f"*{col}_{value} ")
            # Escreve o texto principal (de preferência em minúsculas)
            text = str(row[text_column]).lower().strip().replace('\n', ' ')
            f.write(f"\n{text}\n")

    print(f"[✔] Arquivo salvo como: {output_file}")
```

```python
# Caminho da planilha
file_path = '../data/Bibliometrix-Export-File-2025-06-15.xlsx'  # ou 'dados.csv'

df = pd.read_excel(file_path)

df.head()
```

```python
# Chamada da função
format_for_iramuteq(df, text_column='TI', metadata_columns=['PY'], output_file='corpus_iramuteq.txt')
```
